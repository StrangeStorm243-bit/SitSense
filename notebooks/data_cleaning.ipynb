Step 1: Importing modules and getting current directory to see where we are at

```python
import pandas as pd
import numpy as np
import glob, os
import os
print(os.getcwd())
```

Step 2: We Load Raw Data

```python
posture_files = glob.glob(r"c:\Users\niran\OneDrive\Desktop\SitSense\data\raw\posture_analysis_rows*.csv")
print(posture_files)
```

Step 3: Standardize Columns and Types in Posture Files

```python
posture_dfs = []
for f in posture_files:
    df = pd.read_csv(f)
    # remove accidental header rows
    df = df[df.columns[~df.columns.str.contains("Unnamed")]]
    df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
    # drop rows that look like duplicate headers
    df = df[df[df.columns[0]] != df.columns[0]]
    posture_dfs.append(df)

# --- combine all posture files ---
posture = pd.concat(posture_dfs, ignore_index=True)
```

```python
print(posture.head())
```

Step 4: Clean posture files and save them

```python
# --- clean types and duplicates ---
posture = posture.drop_duplicates().reset_index(drop=True)
posture.columns = posture.columns.str.strip()
# --- save ---
posture.to_csv(r"c:\Users\niran\OneDrive\Desktop\SitSense\data\processed\posture_analysis_cleaned.csv", index=False)
print(" posture_analysis_cleaned.csv saved.")
```

We repeat Step 3 and Step 4 to Clean sessions files

```python
sessions = pd.read_csv(r"c:\Users\niran\OneDrive\Desktop\SitSense\data\raw\sessions_rows.csv")

# clean columns
sessions.columns = sessions.columns.str.strip().str.lower().str.replace(" ", "_")

# remove duplicates or empty rows
sessions = sessions.dropna(how="all").drop_duplicates()

sessions.to_csv(r"c:\Users\niran\OneDrive\Desktop\SitSense\data\processed\sessions_cleaned.csv", index=False)
print(" sessions_cleaned.csv saved.")
```
